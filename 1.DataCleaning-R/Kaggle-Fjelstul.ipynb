The accents seem to be a relevant issue across a lot of players. Let's write a function to normalize all text by removing accents. Also making sure to decode all text in the process.

```r
# This was done with significant help from Codex AI
decode_unicode_tags <- function(x) {
  fix_one <- function(s) {
    matches <- regmatches(
      s,
      gregexpr("<U\\+[0-9A-Fa-f]{4,6}>", s, perl = TRUE)
    )[[1]]

    if (length(matches) == 1 && matches[1] == "") return(s)

    for (m in matches) {
      hex <- sub("<U\\+([0-9A-Fa-f]{4,6})>", "\\1", m, perl = TRUE)
      replacement <- intToUtf8(strtoi(hex, base = 16))
      s <- sub(m, replacement, s, fixed = TRUE)
    }

    s
  }

  vapply(x, fix_one, character(1))
}

remove_accents <- function(x) {
  x <- decode_unicode_tags(x)
  x <- enc2utf8(x)
  stringi::stri_trans_general(x, "Latin-ASCII")
}

#Test
remove_accents("Ronald Araújo")
remove_accents("Ángel Di María")
remove_accents("Cristiano Ronaldo")
```

Seems to work pretty well, lets apply it to both of our datasets and then compare again

I want to assses the relevance of this unmatched players to our dataset and start considering the possibility of excluding some of these players.

```r
nonmatched <- setdiff(worldcup_players$full_name, player_names$name)
fullworldcup_players <- readRDS(here("1.DataCleaning-R", "Data", "RDS", "FullRoster.rds"))

fullworldcup_players %>% 
    filter(full_name %in% nonmatched) %>%
        group_by(team_name, tournament_id) %>%
            summarize(players= sum(n())) %>%
                arrange(desc(players))


```

Looking at this table we can see two clear trends. Firstly, some countries have almost all their players missing, there's a big chance that some of this countries leagues are simply not in our dataste and hence most of their national team players arent available. The other trend is that the oldest the tournamnet the most missing players. Since there is a good chance many of this players we just dont have data for I've decided to take all temas with 8 or more missing players out of our training data. 

```r
fullworldcup_players %>% 
  filter(full_name %in% nonmatched) %>%
    group_by(team_name, tournament_id) %>%
        summarize(players = n(), .groups = "drop") %>%
            filter(players >= 8) %>%
                summarize(total_players = sum(players))

563-265
```

Now we only have 298 missing players, this is reasonable at around 10% of total players. It's inevitable to work with some incomplete data and assuming most of the missing plkayers are lower profile players we can assume that there effect on team performances would be lower so the model shouldn't be affected too strongly.

```r
matched_players <- worldcup_players %>%
  rename(player_id_fj = player_id) %>%
  mutate(wc_year = as.integer(sub("WC-", "", tournament_id))) %>%
  arrange(full_name, desc(wc_year)) %>%
  distinct(full_name, .keep_all = TRUE) %>%
  inner_join(
    player_names %>%
      rename(
        player_id_kag = player_id,
        full_name = name
      ),
    by = "full_name"
  ) %>%
  filter(wc_year <= last_season) %>%
  distinct(full_name, .keep_all = TRUE) %>%
  select(full_name, player_id_fj, player_id_kag)
```
