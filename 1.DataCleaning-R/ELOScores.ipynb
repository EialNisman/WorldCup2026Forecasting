# ELO Scores

## Objectives

Develop a dataset with ELO Scores for each nation in the 2010, 2014, 2018, 2022, and 2026 World Cups. I'm using the ELO scores on Dec 31 the year prior to the world cup. This decision is based on accesibility and consistency. I could not find a reliable way to exctract all ELO scores for each nation at the same date for every world cup at any closer date. I believe it is both a more honest approach and more reliable to set a standard date for all world cups than just using the best date I could find for each tournament. I also compared the ELO scores manually between Dec 31 and datys before the world cup and the differences are minimal.
- Columns: team, WC_2026, WC_2022, WC_2018, WC_2014, WC_2010.
- Rows: One per team that has participated in any of the last 4 world cups.

## Inputs

- `1.DataCleaning-R/Data/RDS/FullTeamGames.rds`
- `1.DataCleaning-R/Data/CSV/ELOStartRatings.csv`
-`1.DataCleaning-R/Data/CSV/ELOAttackDefence.csv`

## Output

- `1.DataCleaning-R/Data/RDS/ELOScores.rds`
- `1.DataCleaning-R/Data/RDS/ATTvsDEF.rds`
