# Players per team per tournament

## Objectives 

Develop a structured dataset with a comprehensive list of all players that were called to each team per tournament.

- Columns: tournament_id, team_name, team_id, team_code, player_id, family_name, given_name, position_code.
- Rows: One per player per tournament (meaning players may be repeated if they participated in multiple world cups).

## Packages

```r
library(tidyverse)
library(worldcup)
library(here)
```

## Rosters

```r
FullRoster <- worldcup::squads
colnames(FullRoster)
```

```r
FullRoster <- FullRoster %>%  
    filter(as.numeric(str_sub(tournament_id, -4)) >= 2010) %>%
        filter(!(tournament_id %in% c("WC-2015", "WC-2019", "WC-2011")))

dim(FullRoster)
```

The dataset description doesnt tell us the specifics of the squads. I'm interested in knowing wether each team had the same amount of players.


```r
FullRoster %>% group_by(tournament_id, team_name) %>%
    summarise(players = sum(n())) %>%
        slice_sample(n=3)
```

We can see 2010-2018 tournaments had 23 players on each squad while 2022 had 26.

```r
FullRoster <- FullRoster %>% 
    select(tournament_id, team_name, team_id, team_code, player_id, family_name, given_name, position_code)
```
