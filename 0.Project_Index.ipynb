# WC2026Forecast Project Index

## Purpose

This notebook is a lightweight table of contents for the forecasting project. It does not scrape data or run the model; it explains the order of the existing notebooks and the artifacts each one produces.

## Recommended Order

1. `1.DataCleaning-R/WorldCups.ipynb` - match-level World Cup results, locations, dates, and rest days.
2. `1.DataCleaning-R/Rosters.ipynb` - team/player rosters for the tournament windows.
3. `1.DataCleaning-R/ELOScores.ipynb` - team strength priors from ELO ratings.
4. `1.DataCleaning-R/Kaggle-Fjelstul.ipynb` - player identifier matching between Kaggle player data and World Cup rosters.
5. `1.DataCleaning-R/PlayerStats.ipynb` and `1.DataCleaning-R/PlayerSeasonStats.ipynb` - player-level production/matching work.
6. `1.DataCleaning-R/Awards.ipynb` - player award points as roster-strength features.
7. `1.DataCleaning-R/Managers.ipynb` - pre-tournament manager experience features.
8. `1.DataCleaning-R/DistanceFromHome.ipynb` - distance from home-country coordinates to host cities.
9. `1.DataCleaning-R/Weather.ipynb` - match weather enrichment.
10. `2.Model_Selection-R/ExploringRelations.ipynb` - exploratory model checks and first relationships.

## Current Modeling Target

The project is moving toward a match/team-level model for goals scored, likely using Poisson-style regression components. The strongest current spine is match history plus ELO, rest, roster/player quality, manager experience, travel distance, and weather.


## Data Artifacts

Primary RDS outputs currently live in `1.DataCleaning-R/Data/RDS/`. CSV files in `1.DataCleaning-R/Data/CSV/` are useful for transparent inspection and manual checks. Large downloaded Kaggle data is ignored by git, which is the right choice for keeping the repo small.


## Notes Before Modeling

- Keep each feature timestamped relative to the tournament being predicted. Anything known only after a World Cup starts can leak future information.
- Separate reproducible local cleaning from scraping/API notebooks. Cells that download or call external services should be clearly marked and avoided during quick project reruns.
- Add a final modeling dataset notebook once the feature tables are stable. That notebook should perform joins, define one row per team-match, and check for missingness before model selection.
